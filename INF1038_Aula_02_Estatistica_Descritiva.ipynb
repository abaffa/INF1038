# Medidas de Dispersão

### Amplitude Total (A)

```python
range_lifeExp = max(data['lifeExp']) - min(data['lifeExp'])
range_year = max(data['year']) - min(data['year'])
print(range_lifeExp)
print(range_year)
```

### Intervalo-Interquartil (d)

```python
import numpy as np
q1_x = np.percentile(data['lifeExp'], 25, method='midpoint')
q3_x = np.percentile(data['lifeExp'], 75, method='midpoint')
print(q3_x - q1_x)
```

```python
from scipy.stats import iqr

print(iqr(data['lifeExp'], rng=(25,75), interpolation='midpoint'))
```

### Desvio Médio Absoluto

```python
from numpy import mean, absolute
#print(data['lifeExp'].mad())
def mad(data, axis=None):
    return mean(absolute(data - mean(data, axis)), axis)

print(mad(data['lifeExp']))    
```

##### Desvio Mediano absoluto
######  o desvio mediano absoluto (MAD) é uma medida de dispersão estatística e é uma medida robusta da variabilidade de uma amostra univariada de dados quantitativos.

```python
from scipy.stats import median_abs_deviation
print(median_abs_deviation(data['lifeExp']))  
```

### Coeficiente de Variação (CV)

```python
data_mean = data['lifeExp'].mean()
print(data_mean)
import statistics as st
print(st.pstdev(data['lifeExp'])/data_mean)

import numpy as np
print(np.std(data['lifeExp'])/data_mean)
```

```python
data_mean = data['lifeExp'].mean()

print(data['lifeExp'].std()/data_mean)

import statistics as st
print(st.stdev(data['lifeExp'])/data_mean)
```

### Resumo dos Cinco Números

```python
from numpy import percentile

quartis = percentile(data['lifeExp'], [25, 50, 75])
data_min, data_max = data['lifeExp'].min(), data['lifeExp'].max()

print('Min: %.3f' % data_min)
print('Q1: %.3f' % quartis[0])
print('Mediana: %.3f' % quartis[1])
print('Q3: %.3f' % quartis[2])
print('Max: %.3f' % data_max)
```

### Identificação de Valores Discrepantes

```python
from numpy import percentile

quartis = percentile(data['lifeExp'], [25, 50, 75])
Q1 = quartis[0]
Q2 = quartis[1]
Q3 = quartis[2]
AIQ = Q3 - Q1

LS = Q3 + 1.5 * AIQ
LI = Q1 - 1.5 * AIQ

print("Limite Inferior: %.3f" % LI)
print("Min: %.3f" % data['lifeExp'].min())
print("Q1: %.3f" % Q1)
print("Média: %.3f" % data['lifeExp'].mean())
print("Mediana: %.3f" % Q2)
print("Q3: %.3f" % Q3)
print("Max: %.3f" % data['lifeExp'].max())
print("Limite Superior: %.3f" % LS)
```
